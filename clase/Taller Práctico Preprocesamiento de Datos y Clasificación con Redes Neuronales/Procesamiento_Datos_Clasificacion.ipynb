## Sección 2: Seleccionar y renombrar columnas

Seleccionamos solo las variables predictoras edad y colesterol, y renombramos:
- `edad` → `x`
- `colesterol` → `y`
- `problema_cardiaco` → `label`

```python
# Seleccionar solo las columnas necesarias
df_selected = df[['edad', 'colesterol', 'problema_cardiaco']].copy()

# Renombrar columnas
df_selected.columns = ['x', 'y', 'label']

print("Dataset después de seleccionar y renombrar columnas:")
print(df_selected.head(10))
print(f"\nForma: {df_selected.shape}")
print(f"\nColumnas: {df_selected.columns.tolist()}")
```

## Sección 3: Revisar y limpiar valores faltantes

Revisamos si existen valores nulos o vacíos en x, y y label, y eliminamos filas con datos faltantes para evitar errores en la estandarización y en el mapeo del target.

```python
# Revisar valores faltantes
print("Valores faltantes antes de limpiar:")
print(df_selected.isnull().sum())
print(f"\nTotal de filas con valores faltantes: {df_selected.isnull().any(axis=1).sum()}")

# Eliminar filas con valores faltantes
df_clean = df_selected.dropna()

print(f"\nForma antes de limpiar: {df_selected.shape}")
print(f"Forma después de limpiar: {df_clean.shape}")
print(f"Filas eliminadas: {df_selected.shape[0] - df_clean.shape[0]}")

print("\nPrimeras 10 filas después de limpiar:")
print(df_clean.head(10))
```

## Sección 4: Mapear la variable objetivo

Transformamos la variable `label` para que:
- La clase positiva `1` se conserve como `1`
- La clase negativa `0` se convierta en `-1`

Esto cumple con el requerimiento técnico de la práctica para clasificación binaria asimétrica.

```python
# Mapear la variable objetivo: 1 → 1 (positivo), 0 → -1 (negativo)
df_clean['label'] = df_clean['label'].map({1: 1, 0: -1})

print("Dataset después de mapear la variable objetivo:")
print(df_clean.head(10))

print(f"\nValores únicos en label: {sorted(df_clean['label'].unique())}")
print(f"Distribución de clases:")
print(df_clean['label'].value_counts().sort_index())

# Verificar que no hay valores faltantes después del mapeo
print(f"\nValores faltantes después del mapeo:")
print(df_clean.isnull().sum())
```

## Sección 5: Estandarizar las features con Z-score y multiplicar por 2

Aplicamos la estandarización Z-score a las columnas `x` e `y` usando la fórmula:
$$z = \frac{x - \text{media}}{\text{desviación estándar}}$$

Luego multiplicamos los valores resultantes por 2 para cumplir con la especificación técnica requerida.

```python
# Calcular media y desviación estándar para x e y
x_mean = df_clean['x'].mean()
x_std = df_clean['x'].std()
y_mean = df_clean['y'].mean()
y_std = df_clean['y'].std()

print("Estadísticas de las features antes de estandarizar:")
print(f"\nx (edad):")
print(f"  Media: {x_mean:.4f}")
print(f"  Desviación Estándar: {x_std:.4f}")
print(f"  Min: {df_clean['x'].min():.4f}")
print(f"  Max: {df_clean['x'].max():.4f}")

print(f"\ny (colesterol):")
print(f"  Media: {y_mean:.4f}")
print(f"  Desviación Estándar: {y_std:.4f}")
print(f"  Min: {df_clean['y'].min():.4f}")
print(f"  Max: {df_clean['y'].max():.4f}")

# Aplicar estandarización Z-score y multiplicar por 2
df_clean['x'] = ((df_clean['x'] - x_mean) / x_std) * 2
df_clean['y'] = ((df_clean['y'] - y_mean) / y_std) * 2

print("\n" + "="*60)
print("Estadísticas de las features después de estandarizar y multiplicar por 2:")
print("="*60)
print(f"\nx (edad estandarizada):")
print(f"  Media: {df_clean['x'].mean():.4f}")
print(f"  Desviación Estándar: {df_clean['x'].std():.4f}")
print(f"  Min: {df_clean['x'].min():.4f}")
print(f"  Max: {df_clean['x'].max():.4f}")

print(f"\ny (colesterol estandarizado):")
print(f"  Media: {df_clean['y'].mean():.4f}")
print(f"  Desviación Estándar: {df_clean['y'].std():.4f}")
print(f"  Min: {df_clean['y'].min():.4f}")
print(f"  Max: {df_clean['y'].max():.4f}")
```

## Sección 6: Validar la salida final

Mostramos el DataFrame final con las columnas x, y y label, verificamos que los tipos de datos sean correctos y confirmamos que los valores estandarizados se hayan multiplicado por 2 y que el target esté mapeado a 1 y -1.

```python
print("="*60)
print("DATASET FINAL PROCESADO")
print("="*60)
print("\nPrimeras 15 filas del dataset procesado:")
print(df_clean.head(15))

print(f"\n\nÚltimas 5 filas del dataset procesado:")
print(df_clean.tail(5))

print(f"\n\nInformación general del dataset:")
print(f"Forma final: {df_clean.shape}")
print(f"Columnas: {df_clean.columns.tolist()}")

print(f"\n\nTipos de datos:")
print(df_clean.dtypes)

print(f"\n\nVerificación del mapeo de target:")
print(f"Valores únicos en label: {sorted(df_clean['label'].unique())}")
print(f"Distribución de clases:")
print(df_clean['label'].value_counts().sort_index())
print(f"Porcentaje de clase positiva (1): {(df_clean['label'] == 1).sum() / len(df_clean) * 100:.2f}%")
print(f"Porcentaje de clase negativa (-1): {(df_clean['label'] == -1).sum() / len(df_clean) * 100:.2f}%")

print(f"\n\nEstadísticas descriptivas finales:")
print(df_clean.describe())

print(f"\n\nVerificación de valores faltantes:")
print(df_clean.isnull().sum())

# Guardar el dataset procesado
df_clean.to_csv('datos_procesados.csv', index=False)
print("\n✓ Dataset procesado guardado en 'datos_procesados.csv'")
```

Convertimos el DataFrame procesado a un formato JSON compatible con un entorno visual. Cada registro contiene las features estandarizadas (x, y) y la variable objetivo mapeada (label con valores 1 o -1).
