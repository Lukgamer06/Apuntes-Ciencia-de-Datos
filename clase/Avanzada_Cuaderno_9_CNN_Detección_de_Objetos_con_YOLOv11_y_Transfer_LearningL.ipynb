La detección de objetos es una tarea de visión por computador que no solo identifica qué objetos hay en una imagen, sino también dónde están ubicados. Combina dos elementos:

**Clasificación** (¿qué es?),

**Localización** (¿dónde está?).

Red de detección de objetos aplicada a una foto puede identificar un perro, una pelota, y dibujr cuadros delimitadores (bounding boxes) alrededor de cada uno.

La detección de objetos es una tarea de visión por computador que combina:

**Regresión:** para predecir la posición y tamaño de cada objeto en la imagen, generalmente en forma de (x, y, ancho, alto).

**Clasificación:** para identificar qué clase representa cada objeto detectado (por ejemplo: persona, perro, auto).

Cada modelo de detección realiza ambas tareas simultáneamente.

Diferencia entre detección de objetos y reconocimiento de imágenes
Reconocimiento de imágenes	Detección de objetos


* **Reconocimiento de Imágenes:** Clasifica toda la imagen. Una sola etiqueta por imagen: "Esto es un perro"

* **Detección de objetos:** Detecta y localiza múltiples objetos. Múltiples objetos y clases por imagen: "Aquí hay un proo y aquí un balón"

### Funciones para descargar y extraer el dataset

Se definen dos funciones:
- `download_zip`: descarga el archivo desde GitHub.
- `extract_zip`: extrae el archivo ZIP en la carpeta `dataset/`.

Después se ejecutan ambas funciones.


````python
def download_zip(url: str, destination: Path) -> None:
    destination.parent.mkdir(parents=True, exist_ok=True)
    response = requests.get(url, stream=True, timeout=60)
    response.raise_for_status()
    with destination.open("wb") as file:
        for chunk in response.iter_content(chunk_size=8192):
            if chunk:
                file.write(chunk)
    print(f"Dataset descargado: {destination}")


def extract_zip(zip_file: Path, destination: Path) -> None:
    if not zip_file.is_file():
        raise FileNotFoundError(f"No se encontró el archivo ZIP: {zip_file}")
    destination.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(zip_file, "r") as archive:
        archive.extractall(destination)
    print(f"Dataset extraído en: {destination}")
    print(f"Contenido: {sorted(path.name for path in destination.iterdir())}")


download_zip(
    "https://github.com/adiacla/bigdata/blob/master/segmentation.v1i.yolov8.zip?raw=true",
    DATASET_ZIP,
)
extract_zip(DATASET_ZIP, DATASET_DIR)
````

### Implementación de Funciones de Descarga y Extracción

Se definen dos funciones esenciales para automatizar la gestión del dataset:

*   `download_zip(url, dest)`: Descarga un archivo ZIP desde una URL específica y lo guarda en la ruta indicada. Utiliza `requests` para una descarga robusta.
*   `extract_zip(zip_file, dest_folder)`: Descomprime el archivo ZIP descargado en la carpeta `dataset/` dentro del directorio de destino. Esto organiza los archivos del dataset en la estructura esperada por YOLO11.

Estas funciones son invocadas para obtener y preparar el dataset en el entorno de Colab.

### ntrenamiento del modelo YOLO11 con dataset de Roboflow

Ya puedes usar el archivo `data.yaml` para entrenar tu modelo con:


````python
data_yaml = DATASET_DIR / "data.yaml"
if not data_yaml.is_file():
    raise FileNotFoundError(f"No se encontró la configuración del dataset: {data_yaml}")

with data_yaml.open("r", encoding="utf-8") as file:
    dataset_config = yaml.safe_load(file)
if not isinstance(dataset_config, dict):
    raise ValueError(f"El archivo no contiene una configuración YAML válida: {data_yaml}")
dataset_config["path"] = str(DATASET_DIR)
with data_yaml.open("w", encoding="utf-8") as file:
    yaml.safe_dump(dataset_config, file, sort_keys=False, allow_unicode=True)

model.train(
    data=str(data_yaml),
    epochs=50,
    imgsz=640,
    batch=16,
    project=str(RUNS_DIR),
    name="yolo11_ppe",
    exist_ok=True,
)
# Ajusta batch según la memoria disponible en tu GPU.
````

### Entrenamiento del Modelo con el Dataset Personalizado

El método `model.train()` inicia el proceso de *fine-tuning* de nuestro modelo YOLO11. Se configuran parámetros esenciales como:

*   `data`: La ruta al archivo `data.yaml` que define el dataset.
*   `epochs`: El número de veces que el modelo revisará el dataset completo (aquí, 50 épocas).
*   `imgsz`: El tamaño de la imagen de entrada para el modelo (640x640 píxeles).
*   `batch`: El número de imágenes procesadas en cada iteración (aquí, 16 imágenes).

Estos parámetros son cruciales para optimizar el rendimiento y la velocidad del entrenamiento.

## 9.3 Aplicación de detección con Streamlit

La app carga exclusivamente el `best.pt` que acaba de entrenarse. Permite subir una imagen, ajustar el umbral de confianza y consultar las detecciones. El archivo de modelo se genera en este entorno, no se solicita a través de la interfaz.

Ejecuta esta celda después de entrenar y copiar `best.pt`.

````python
from pathlib import Path


deployment_dir = PROJECT_DIR / "deployment_yolo11"
deployment_dir.mkdir(parents=True, exist_ok=True)

app_source = r'''from pathlib import Path

import streamlit as st
from PIL import Image, UnidentifiedImageError
from ultralytics import YOLO

APP_DIR = Path(__file__).resolve().parent
MODEL_PATH = APP_DIR / "best.pt"
MAX_UPLOAD_BYTES = 10 * 1024 * 1024
MAX_IMAGE_PIXELS = 20_000_000

st.set_page_config(page_title="Detector YOLO11", page_icon="🔎", layout="centered")
st.title("Detección de objetos con YOLO11")
st.write("Sube una imagen para detectar los objetos del modelo entrenado.")


@st.cache_resource
def load_model(model_path: str) -> YOLO:
    return YOLO(model_path)


if not MODEL_PATH.is_file():
    st.error("No se encontró best.pt. Ejecuta primero el cuaderno de entrenamiento.")
    st.stop()

confidence = st.slider("Confianza mínima", 0.05, 0.95, 0.25, 0.05)
uploaded_file = st.file_uploader(
    "Selecciona una imagen",
    type=["jpg", "jpeg", "png", "webp"],
)

if uploaded_file is not None:
    if uploaded_file.size > MAX_UPLOAD_BYTES:
        st.error("La imagen supera el límite de 10 MB.")
        st.stop()

    try:
        with Image.open(uploaded_file) as source_image:
            if source_image.width * source_image.height > MAX_IMAGE_PIXELS:
                st.error("La imagen supera el límite de 20 megapíxeles.")
                st.stop()
            image = source_image.convert("RGB")
    except (UnidentifiedImageError, OSError, Image.DecompressionBombError):
        st.error("El archivo no contiene una imagen válida o no se puede decodificar.")
        st.stop()

    model = load_model(str(MODEL_PATH))
    with st.spinner("Analizando imagen..."):
        result = model.predict(source=image, imgsz=640, conf=confidence, verbose=False)[0]

    annotated_image = result.plot()[:, :, ::-1]
    st.image(annotated_image, caption="Resultado de la detección", use_container_width=True)

    if result.boxes is not None and len(result.boxes) > 0:
        detections = []
        for box in result.boxes:
            class_id = int(box.cls.item())
            detections.append({
                "Clase": result.names[class_id],
                "Confianza": round(float(box.conf.item()), 3),
            })
        st.dataframe(detections, use_container_width=True)
    else:
        st.info("No se detectaron objetos con ese umbral de confianza.")
'''

compile(app_source, "deployment_yolo11/app.py", "exec")
(deployment_dir / "app.py").write_text(app_source, encoding="utf-8")
(deployment_dir / "requirements.txt").write_text(
    "streamlit>=1.36.0\nultralytics>=8.3.0\nPillow>=10.0.0\n",
    encoding="utf-8",
)
config_dir = deployment_dir / ".streamlit"
config_dir.mkdir(exist_ok=True)
(config_dir / "config.toml").write_text(
    "[server]\nmaxUploadSize = 10\n",
    encoding="utf-8",
)
print(f"Aplicación y dependencias creadas en: {deployment_dir}")
````

La carpeta `deployment_yolo11` se crea dentro de la carpeta del proyecto abierta en VS Code. Incluye `app.py`, `requirements.txt`, `.streamlit/config.toml` y `best.pt`.
