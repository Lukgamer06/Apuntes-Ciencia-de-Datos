La detección de objetos es una tarea de visión por computador que no solo identifica qué objetos hay en una imagen, sino también dónde están ubicados. Combina dos elementos:

**Clasificación** (¿qué es?),

**Localización** (¿dónde está?).

*Ejemplo:*

Red de detección de objetos aplicada a una foto puede identificar un perro, una pelota, y dibujr cuadros delimitadores (bounding boxes) alrededor de cada uno.

La detección de objetos es una tarea de visión por computador que combina:

**Regresión:** para predecir la posición y tamaño de cada objeto en la imagen, generalmente en forma de (x, y, ancho, alto).

**Clasificación:** para identificar qué clase representa cada objeto detectado (por ejemplo: persona, perro, auto).

Cada modelo de detección realiza ambas tareas simultáneamente.

Diferencia entre detección de objetos y reconocimiento de imágenes
Reconocimiento de imágenes	Detección de objetos


* **Reconocimiento de Imágenes:** Clasifica toda la imagen. Una sola etiqueta por imagen: "Esto es un perro"

* **Detección de objetos:** Detecta y localiza múltiples objetos. Múltiples objetos y clases por imagen: "Aquí hay un proo y aquí un balón"

### Funciones para descargar y extraer el dataset

Se definen dos funciones:
- `download_zip`: descarga el archivo desde GitHub.
- `extract_zip`: extrae el archivo ZIP en la carpeta `dataset/`.

Después se ejecutan ambas funciones.


````python
# Definir rutas base
base_path = "/content"
zip_path = os.path.join(base_path, "segmentation.v1i.yolov8.zip")

def download_zip(url, dest):
    response = requests.get(url, stream=True)
    with open(dest, "wb") as file:
        for chunk in response.iter_content(chunk_size=8192):
            file.write(chunk)
    print("Archivo ZIP descargado correctamente.")

def extract_zip(zip_file, dest_folder):
    dataset_folder = os.path.join(dest_folder, "dataset")  # Extraer en "dataset/"
    os.makedirs(dataset_folder, exist_ok=True)  # Asegurar que la carpeta existe
    with zipfile.ZipFile(zip_file, 'r') as zip_ref:
        zip_ref.extractall(dataset_folder)  # Extraer en dataset/
    print(f"Contenido extraído en {dataset_folder}: {os.listdir(dataset_folder)}")


# Ejecutar procesos
download_zip("https://github.com/adiacla/bigdata/blob/master/segmentation.v1i.yolov8.zip?raw=true", zip_path)
extract_zip(zip_path, base_path)
````

### Implementación de Funciones de Descarga y Extracción

Se definen dos funciones esenciales para automatizar la gestión del dataset:

*   `download_zip(url, dest)`: Descarga un archivo ZIP desde una URL específica y lo guarda en la ruta indicada. Utiliza `requests` para una descarga robusta.
*   `extract_zip(zip_file, dest_folder)`: Descomprime el archivo ZIP descargado en la carpeta `dataset/` dentro del directorio de destino. Esto organiza los archivos del dataset en la estructura esperada por YOLOv8.

Estas funciones son invocadas para obtener y preparar el dataset en el entorno de Colab.

### Entrenamiento desde Cero (Scratch) vs. Transfer Learning

Esta celda ilustra la diferencia entre entrenar un modelo desde cero y usar *transfer learning*.

*   **Entrenar desde cero (`model = YOLO("yolov8s.yaml")`):** Implica construir un modelo sin ningún conocimiento previo. Es útil cuando se tienen datasets muy grandes y muy específicos, pero requiere más tiempo y recursos computacionales.
*   **Transfer Learning (`model = YOLO("/runs/detect/train/weights/last.pt")`):** Consiste en cargar un modelo preentrenado (como el que hicimos) y ajustarlo a un nuevo dataset. Es el enfoque recomendado en la mayoría de los casos, ya que acelera el entrenamiento y mejora el rendimiento, especialmente con datasets más pequeños. La ruta `/runs/detect/train/weights/last.pt` apunta a los pesos del último entrenamiento realizado.
